### Overall conclusions about reproducibility: 
We found that the reproducibility of this data to be... medium. We rated it from 0-5 stars, where 0 was “impossible to reproduce” and 5 was “seamless and even kinda fun.” Jeanne rated it at a 2, Avery at a 2.4, and Michael at a generous 3. The average and overall score was 2.5 stars. 

It was possible to reproduce the methods, but we found different results in our figures. This indicates to us that the biological signal is not robust to different bioinformatic techniques. There were times when we choose to use different methods than the paper, usually when the technique was out-of-date or overly complicated. We also had to make several decisions on our own, where the paper was unclear or did not include methodological descriptions. 

Given the challenges of developing our own pipeline for the data and the notable differences between the figure we created and that of the paper, we gave it a low reproducibility score.

Overall, I was surprised at how different the two figures look from one another. It is important to note, however, that our figure only looks at the four vents from Piccard. However, there are very abundances in hydrogen, oxygen, carbon, and sulfur related genes at Hot Chimlet in the original figure that are not captured in our reproduction. Our data are relatively constant between vents, with much less variation than the original figure. More encouragingly, we were able to recreate the low abundance of methane and nitrogen related genes that is present in the original data. 

I am not sure why our differences occur, but the most suspicious part of our pipeline was when 100% of the reads mapped to the assembled genome. This makes me wonder if the data on ENA was not truly raw. Perhaps it was accidentally manipulated in a way that made it too clean and caused a loss in some of the variability amongst samples. The paper did not specify if the reads were mapped to a concatenated or individual assemblies. We chose concatenated assemblies. Perhaps this is what the number of mapped reads is more even across samples.

#### Response (Avery)
In the paper, the authors concluded that they saw no clear differences in the transcriptional abundance of metabolic genes of interest between ultramafic and mafic systems. They did note that there appears to be more transcription of hydrogenases and mcr in ultramafic systems, consistent with the higher geochemical availability of H2 and higher rates of methanogenesis in these systems. 

While the figures are different, I think it is important to note that our reproduced figure could have led to the same conclusions as discussed in the paper. Of course, we did not replicate the ultramafic samples, but there appears to be a similar relative distribution of transcriptional activity between each group of metabolic genes of interest in the mafic samples. In particular, there are low levels of transcriptional activity for mcr, and the hydrogenases appear to also be transcribed at a medium level in comparison to the other metabolic genes within the mafic sample set. 

Yet our reproduced figure is notably different, particularly when it comes to the evenness of transcriptional abundance between sample sites. At several mafic sites, the authors of the paper found little to no transcripts in several key metabolic categories. In comparison, we recovered all metabolic genes of interest from almost all sample sites. I wonder if this is because, at Harriet's suggestion, we skipped the step of "manually checking ORFs of interest" with tblastn searches using an e-value cutoff of 10−12. Our overall transcript abundance is 1-2 orders of magnitude larger, based on their figure legend, and I wonder if this is because the authors determined that some of the genes were incorrectly annotated with KEGG orthology as genes of interest. 

The other possible source of divergence between our figures could be caused by the very confusing (in my opinion) methods section describing how the normalized gene abundance was calculated. In the methods, they write, "If multiple ORFs had the same annotation, the number of read hits was averaged. ... To calculate the transcript abundance of each gene, the number of metatranscriptomic read hits to each key gene was divided by the total number of metatranscriptomic reads. Again, read hits for ORFs with the same annotation were averaged." The authors really wanted us to average read hits for ORFs with the same annotation, but I was unsure about exactly how they did this — did they average ORFs with the same gene annotation, or KO ID? If ORFs were part of the same gene complex (mcrA vs mcrB, for example), did they average these together or separately? Upon asking Harriet, Arianna, and Jeanne, I only averaged read hits for single-copy COGs, but this could have been a source of difference. 

Of course, the different read hit abundances could also have been caused by divergences in our pipeline, such as the fact that we used salmon instead of JGI IMG to count the number of reads that mapped back to ORFs of interest, or because we used megahit instead of idba-ud for the assembly of metagenomic reads and therefore missed subtle alignment differences that would have been caught by a program with higher memory requirements.

#### Response (Michael)

One of the major parts of this paper is the use of module completion ratio, or MCS, a metric used to estimate the functional metabolic potential of different MAGs based on how "complete" a metabolic pathway can be built by the genes within a certain MAG. One main problem with this method is that it cannot distinguish between pathways that use the same set of genes, such as methanogenesis and ANME. Thus one of the goals of Fig 5 is to provide some insight into which mcrA metabolic pathways are being used at different sites.
