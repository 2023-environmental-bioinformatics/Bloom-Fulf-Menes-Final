## First attempt at downloading data to HPC -- authors JB and AF

```python
filename = '../output/download_data/filereport_read_run_PRJEB15541_tsv.txt' 
#this is all of the files in the ENA, including those we do not want
```

```python
with open(filename,'r') as f: #open every filename in its own line
    for line in f:
        print(line)
f.close()
```

```python
with open(filename,'r') as f:
    finalfilenames = [] #make an empty list for the urls to go into
    for line in f:
        l = line.split('\t')[7].split(';') #split the lines by a tab then by the semicolon into columns
        for link in l:
            if 'FS851' in link or 'FS852' in link or 'FS854' in link or 'FS856' in link: 
                #only extract the links with the names of the samples that relate to Piccard vent (from the paper)
                finalfilenames.append(link)
f.close()
```

```python
print(finalfilenames)
```

```python
import os #this package has functions that engages with HPC from within python
```

```python
for url in finalfilenames: #begin loop that extracts the urls from the filename (from above) 
    os.system('wget ' + url) #wget downloads the URL from the ENA (or any url) to the HPC
```
